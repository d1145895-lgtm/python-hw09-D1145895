```python
# HW09 兩場活動報名整理器
- 姓名：林敏妍
- 學號：D1145895

本程式用來整理桌遊體驗與電影欣賞兩場活動的原始報名名單。
先將文字統一轉成小寫並去除空白，接著透過集合剔除單場重複報名的人員並比較兩場參加者，
最後計算各場人數、總不重複人數、報名人次與應收金額。
```

```python
# 固定的活動名稱與單價設定
activity_names = ("桌遊體驗", "電影欣賞")
unit_fees = {"桌遊體驗": 120, "電影欣賞": 80}

# 原始名單輸入（保留原始變數不被覆蓋）
board_raw = " Amy, Ben, amy, Cora "
movie_raw = " ben, Dora, ERIC, ben "
```

```python
# 1. 讀取活動名稱並整理字串為串列（保留原始順序與重複項目）
board_name, movie_name = activity_names
board_list = board_raw.replace(" ", "").lower().split(",")
movie_list = movie_raw.replace(" ", "").lower().split(",")

# 2. 轉換為集合進行去重與名單比較
board_set = set(board_list)
movie_set = set(movie_list)

both_set = board_set & movie_set          # 交集：兩場都參加
all_set = board_set | movie_set           # 聯集：至少參加一場
only_board_set = board_set - movie_set    # 差集：只參加桌遊
only_movie_set = movie_set - board_set    # 差集：只參加電影

# 3. 排序名單並取得前兩位通知預覽
board_sorted = sorted(board_set)
movie_sorted = sorted(movie_set)
both_sorted = sorted(both_set)
only_board_sorted = sorted(only_board_set)
only_movie_sorted = sorted(only_movie_set)
all_sorted = sorted(all_set)

preview_list = all_sorted[:2]

# 4. 計算人數與金額，存入 summary 字典
board_count = len(board_set)
movie_count = len(movie_set)
board_fee = board_count * unit_fees[board_name]
movie_fee = movie_count * unit_fees[movie_name]

summary = {
    "board_count": board_count,
    "movie_count": movie_count,
    "both_count": len(both_set),
    "unique_total": len(all_set),
    "person_times": board_count + movie_count,
    "board_fee": board_fee,
    "movie_fee": movie_fee,
    "total_fee": board_fee + movie_fee,
}

# 5. 輸出完整摘要
print("活動報名摘要（測試一）")
print(f"活動：{board_name}，每人 {unit_fees[board_name]:.2f} 元")
print(f"活動：{movie_name}，每人 {unit_fees[movie_name]:.2f} 元")
print(f"桌遊整理串列：{board_list}")
print(f"電影整理串列：{movie_list}")
print(f"桌遊去重名單：{board_sorted}")
print(f"電影去重名單：{movie_sorted}")
print(f"桌遊人數：{summary['board_count']} 人")
print(f"電影人數：{summary['movie_count']} 人")
print(f"共同參加者：{both_sorted}")
print(f"共同人數：{summary['both_count']} 人")
print(f"只參加桌遊者：{only_board_sorted}")
print(f"只參加電影者：{only_movie_sorted}")
print(f"全部參加者：{all_sorted}")
print(f"總不重複人數：{summary['unique_total']} 人")
print(f"報名人次：{summary['person_times']} 人次")
print(f"通知名單前兩位：{preview_list}")
print(f"桌遊應收：{summary['board_fee']:.2f} 元")
print(f"電影應收：{summary['movie_fee']:.2f} 元")
print(f"總應收：{summary['total_fee']:.2f} 元")
```
